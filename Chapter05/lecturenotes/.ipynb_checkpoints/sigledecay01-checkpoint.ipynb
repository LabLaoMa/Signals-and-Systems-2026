```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 设置参数
w = np.linspace(-2 * np.pi, 2 * np.pi, 1000)  # 频率范围从 -2pi 到 2pi
a1 = 0.5
a2 = -0.5

# 2. 定义计算幅度和相位的函数
def get_dtft_properties(w, a):
    # 根据图片中的公式计算
    # 幅度: |X(e^jw)| = 1 / sqrt(1 - 2a cos(w) + a^2)
    magnitude = 1 / np.sqrt(1 - 2 * a * np.cos(w) + a**2)
    
    # 相位: arg X(e^jw) = -arctan( (a sin(w)) / (1 - a cos(w)) )
    phase = -np.arctan2(a * np.sin(w), 1 - a * np.cos(w))
    
    return magnitude, phase

# 计算两种情况的数据
mag1, phase1 = get_dtft_properties(w, a1)
mag2, phase2 = get_dtft_properties(w, a2)

# 3. 开始绘图
fig, axs = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)

# --- 第一列: a = 0.5 ---
# 幅度图
axs[0, 0].plot(w, mag1, 'b', lw=2)
axs[0, 0].set_title(r'$X_1(e^{j\omega})$ for $a=0.5$ (Magnitude)', fontsize=14)
axs[0, 0].set_ylabel('Magnitude')

# 相位图
axs[1, 0].plot(w, phase1, 'b', lw=2)
axs[1, 0].set_title(r'$\arg X_1(e^{j\omega})$ (Phase)', fontsize=14)
axs[1, 0].set_ylabel('Phase (radians)')

# --- 第二列: a = -0.5 ---
# 幅度图
axs[0, 1].plot(w, mag2, 'b', lw=2)
axs[0, 1].set_title(r'$X_2(e^{j\omega})$ for $a=-0.5$ (Magnitude)', fontsize=14)

# 相位图
axs[1, 1].plot(w, phase2, 'b', lw=2)
axs[1, 1].set_title(r'$\arg X_2(e^{j\omega})$ (Phase)', fontsize=14)

# 4. 格式化所有子图
for ax in axs.flat:
    ax.axhline(0, color='black', lw=1)
    ax.axvline(0, color='black', lw=1)
    ax.set_xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi])
    ax.set_xticklabels([r'$-2\pi$', r'$-\pi$', '0', r'$\pi$', r'$2\pi$'])
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.set_xlabel(r'$\omega$')

plt.show()
```
