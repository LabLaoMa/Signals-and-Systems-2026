```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def plot_dtft(a):
    # 1. 准备频率轴数据
    w = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
    
    # 2. 计算幅度和相位 (基于图片中的公式)
    # 幅度: 1 / sqrt(1 - 2acos(w) + a^2)
    magnitude = 1 / np.sqrt(1 - 2 * a * np.cos(w) + a**2)
    
    # 相位: -arctan( (asin(w)) / (1 - acos(w)) )
    phase = -np.arctan2(a * np.sin(w), 1 - a * np.cos(w))
    
    # 3. 绘图
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))
    plt.subplots_adjust(hspace=0.4)
    
    # --- 幅频响应图 ---
    ax1.plot(w, magnitude, 'b', lw=2)
    ax1.set_title(f'Magnitude Spectrum |X(e^jω)| (a = {a:.2f})', fontsize=14)
    ax1.set_ylim(0, 1/(1-abs(a)) + 1 if abs(a) < 1 else 10)
    ax1.set_ylabel('Magnitude')
    
    # --- 相频响应图 ---
    ax2.plot(w, phase, 'r', lw=2)
    ax2.set_title(f'Phase Spectrum arg X(e^jω) (a = {a:.2f})', fontsize=14)
    ax2.set_ylabel('Phase (radians)')
    
    # 格式化坐标轴
    for ax in [ax1, ax2]:
        ax.axhline(0, color='black', lw=0.8)
        ax.axvline(0, color='black', lw=0.8)
        ax.set_xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi])
        ax.set_xticklabels([r'$-2\pi$', r'$-\pi$', '0', r'$\pi$', r'$2\pi$'])
        ax.set_xlabel(r'$\omega$')
        ax.grid(True, linestyle=':', alpha=0.6)

    plt.show()

# 创建交互式滑块
# |a| 必须小于 1 才能保证系统稳定/级数收敛
interact(plot_dtft, a=FloatSlider(min=-0.95, max=0.95, step=0.05, value=0.5, description='a 值:'));
```
