```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

def plot_time_expansion(m=1):
    # 1. 定义原始信号 x[n]
    n_orig = np.arange(-10, 11)
    # 使用 Hamming 窗截断的 sinc 函数，频谱更漂亮
    x_orig = np.sinc(n_orig * 0.4) * np.hamming(len(n_orig))
    
    # 2. 构造时间扩展信号 x_(m)[n]
    x_ext = np.zeros(len(x_orig) * m)
    x_ext[::m] = x_orig
    # 修正中心对齐
    center = (len(x_ext) // 2)
    n_ext = np.arange(len(x_ext)) - (len(x_orig)//2 * m)
    
    # 3. 计算 DTFT
    N_fft = 2048
    X_orig_freq = np.fft.fftshift(np.fft.fft(x_orig, N_fft))
    X_ext_freq = np.fft.fftshift(np.fft.fft(x_ext, N_fft))
    w = np.linspace(-np.pi, np.pi, N_fft)

    fig, axes = plt.subplots(2, 2, figsize=(12, 7))
    plt.subplots_adjust(hspace=0.4, wspace=0.3)

    # --- 时域图 ---
    axes[0, 0].stem(n_orig, x_orig)
    axes[0, 0].set_title("Original Signal $x[n]$")
    axes[0, 0].set_xlim(-25, 25)
    axes[0, 0].grid(True, alpha=0.3)

    axes[1, 0].stem(n_ext, x_ext, linefmt='C1-', markerfmt='C1o')
    axes[1, 0].set_title(f"Time Expanded Signal $x_{(m)}[n]$ (m={m})")
    axes[1, 0].set_xlim(-25, 25)
    axes[1, 0].grid(True, alpha=0.3)

    # --- 频域图 ---
    axes[0, 1].plot(w, np.abs(X_orig_freq))
    axes[0, 1].set_title(r"Magnitude Spectrum $|X(e^{j\omega})|$")
    axes[0, 1].set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
    axes[0, 1].set_xticklabels([r'$-\pi$', r'$-\pi/2$', '0', r'$\pi/2$', r'$\pi$'])
    axes[0, 1].grid(True, alpha=0.3)

    axes[1, 1].plot(w, np.abs(X_ext_freq), color='orange')
    # 使用字符串拼接规避 f-string backslash 错误
    axes[1, 1].set_title(r"Compressed Spectrum $|X(e^{jm\omega})|$ (m=" + str(m) + ")")
    axes[1, 1].set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
    axes[1, 1].set_xticklabels([r'$-\pi$', r'$-\pi/2$', '0', r'$\pi/2$', r'$\pi$'])
    axes[1, 1].grid(True, alpha=0.3)

    plt.show()

interact(plot_time_expansion, m=IntSlider(min=1, max=5, step=1, value=1));
```
