```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# 设置参数
n = 5  # 阶数
rp = 1 # 通带纹波 (dB)
rs = 40 # 阻带衰减 (dB)
wn = 1 # 归一化截止频率

# 创建滤波器
b_butt, a_butt = signal.butter(n, wn, 'low', analog=True)
b_elip, a_elip = signal.ellip(n, rp, rs, wn, 'low', analog=True)

# 计算阶跃响应
t, y_butt = signal.step((b_butt, a_butt))
t, y_elip = signal.step((b_elip, a_elip))

# 绘图
plt.plot(t, y_butt, '--', label='Butterworth filter')
plt.plot(t, y_elip, '-', label='Elliptic filter')
plt.axhline(y=1, color='gray', linestyle=':')
plt.legend(loc='best')
plt.xlabel('Time')
plt.ylabel('Magnitude')
plt.grid(True)
plt.show()
```
