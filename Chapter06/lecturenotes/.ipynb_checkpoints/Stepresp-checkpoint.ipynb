```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from ipywidgets import interact, FloatSlider

def plot_step_response(zeta):
    # 设定自然频率 omega_n 为 1 (归一化)
    wn = 1.0
    
    # 构造二阶系统的传递函数 H(s) = wn^2 / (s^2 + 2*zeta*wn*s + wn^2)
    # 分子系数: [wn^2]
    num = [wn**2]
    # 分母系数: [1, 2*zeta*wn, wn^2]
    den = [1, 2*zeta*wn, wn**2]
    
    sys = signal.TransferFunction(num, den)
    
    # 计算阶跃响应
    t, y = signal.step(sys)
    
    # 绘图
    plt.figure(figsize=(10, 6))
    plt.plot(t, y, lw=2, label=f'$\zeta = {zeta:.2f}$')
    plt.axhline(y=1, color='r', linestyle='--', alpha=0.5, label='稳态值')
    
    # 设置图表细节
    plt.title('二阶系统单位阶跃响应特性', fontsize=15)
    plt.xlabel('时间 $t$', fontsize=12)
    plt.ylabel('响应 $y(t)$', fontsize=12)
    plt.ylim(0, 2.0)
    plt.grid(True, which='both', linestyle=':', alpha=0.7)
    plt.legend(loc='upper right')
    
    # 添加物理特性标注
    if zeta < 1:
        state = "欠阻尼 (Underdamped)"
    elif zeta == 1:
        state = "临界阻尼 (Critically Damped)"
    else:
        state = "过阻尼 (Overdamped)"
    plt.text(t[-1]*0.6, 0.2, f"系统状态: {state}", 
             bbox=dict(facecolor='white', alpha=0.5))
    
    plt.show()

# 创建滑块交互，范围从 0 到 2
interact(plot_step_response, 
         zeta=FloatSlider(min=0.01, max=2.0, step=0.05, value=0.707, description='阻尼比 $\zeta$:'));
```
